```python
# covid_india_viz_2020_2021.py
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import json

# Create folders
os.makedirs("data", exist_ok=True)
os.makedirs("figures", exist_ok=True)


# ---- 1) Obtain the statewise data file (one of these sources) ----
# Preferred file: https://data.covid19india.org/csv/latest/state_wise.csv
# Alternative for daily: https://data.covid19india.org/csv/latest/state_wise_daily.csv
# If automatic download fails, download manually and save as data/state_wise.csv


df_states = pd.read_csv(r"C:\Users\Dhiraj Mazumdar\Documents\python\covid_data\data_state_wise.csv")
df_states.head()

# Filter relevant columns if present
# Expected typical columns: State/UT, Confirmed, Recovered, Deaths, Active
# Normalize column names:
```

```python
# Filter relevant columns if present
# Expected typical columns: State/UT, Confirmed, Recovered, Deaths, Active
# Normalize column names:

df_states.columns = [c.strip() for c in df_states.columns]
print("Columns:", df_states.columns.tolist())
```

```python
# Keep only State/UT rows (some files include totals or area rows)
# Assuming first column is 'State' or 'State/UT' or 'State/ut'
state_col_candidates = [c for c in df_states.columns if 'state' in c.lower()]
if state_col_candidates:
    state_col = state_col_candidates[0]
else:
    # fallback to first column
    state_col = df_states.columns[0]
print("Using state column:", state_col)

```

```python
# --- CLEANUP AND COLUMN NORMALIZATION BLOCK (fixed version) ---

# Standardize column names to lowercase and strip spaces
df_states.columns = df_states.columns.str.strip().str.lower()

# Try to detect the state column automatically
state_col_candidates = [c for c in df_states.columns if 'state' in c or 'ut' in c]
if len(state_col_candidates) == 0:
    raise ValueError("❌ No column found that looks like a state name column!")
state_col = state_col_candidates[0]

# Map columns by best match
rename_map = {}
for c in df_states.columns:
    if 'confirm' in c:
        rename_map[c] = 'Confirmed'
    elif 'recover' in c:
        rename_map[c] = 'Recovered'
    elif 'death' in c:
        rename_map[c] = 'Deaths'
    elif 'active' in c:
        rename_map[c] = 'Active'

# Rename them to standard form
df_states = df_states.rename(columns=rename_map)

# Keep only necessary columns (if present)
cols_to_keep = ['Confirmed', 'Recovered', 'Deaths', 'Active']
existing = [c for c in cols_to_keep if c in df_states.columns]
if not existing:
    raise ValueError("❌ Could not find numeric columns for Confirmed/Recovered/Deaths/Active. Check CSV header names.")

# Filter out total / unassigned rows
df_states = df_states[
    ~df_states[state_col].str.lower().isin(['india', 'total', 'tt', 'state unassigned'])
]

# Subset DataFrame
df = df_states[[state_col] + existing].copy()
df = df.rename(columns={state_col: 'State'})

# Convert to numeric safely
for c in existing:
    df[c] = pd.to_numeric(df[c], errors='coerce').fillna(0).astype(int)

# Compute derived metrics
if all(col in df.columns for col in ['Confirmed','Deaths']):
    df['CFR'] = (df['Deaths'] / df['Confirmed'] * 100).round(2)
if all(col in df.columns for col in ['Confirmed','Active']):
    df['ActivePct'] = (df['Active'] / df['Confirmed'] * 100).round(2)

# Sort states by confirmed cases
df = df.sort_values('Confirmed', ascending=False).reset_index(drop=True)

print("✅ Cleaned data:")
print(df.head(10))
print(df.columns)
print(df.head())
```

```python
# --- Consolidate duplicated columns (collapse duplicates into single columns) ---
import numpy as np

print("Before consolidation, columns:", df.columns.tolist())

consolidated = df.copy()

# iterate unique column names in original order
unique_cols = []
for col in consolidated.columns:
    if col not in unique_cols:
        unique_cols.append(col)

for col in unique_cols:
    # select all columns with this name (this returns a DataFrame if duplicates exist)
    same = consolidated.loc[:, consolidated.columns == col]
    if same.shape[1] > 1:
        # Combine by taking first non-null value across columns (row-wise)
        # bfill fills NaNs to the right, so column 0 becomes first non-null
        combined = same.bfill(axis=1).iloc[:, 0]
        consolidated[col] = combined
        print(f"Consolidated {same.shape[1]} duplicate columns for '{col}' -> single column")
    else:
        # single column, nothing to do
        pass

# Drop duplicate columns (keep first occurrence)
consolidated = consolidated.loc[:, ~consolidated.columns.duplicated()]

print("After consolidation, columns:", consolidated.columns.tolist())

# Now normalize names (lowercase, strip) if you want, or proceed to rename
# Example: ensure State column name is standardized
# find state column
state_candidates = [c for c in consolidated.columns if 'state' in c.lower()]
if not state_candidates:
    raise ValueError("No state-like column found after consolidation. Columns: " + str(consolidated.columns.tolist()))
state_col = state_candidates[0]

# Rename numeric columns to standard names if needed
rename_map = {}
for c in consolidated.columns:
    lc = c.lower()
    if 'confirm' in lc:
        rename_map[c] = 'Confirmed'
    elif 'recover' in lc:
        rename_map[c] = 'Recovered'
    elif 'death' in lc:
        rename_map[c] = 'Deaths'
    elif 'active' in lc:
        rename_map[c] = 'Active'

consolidated = consolidated.rename(columns=rename_map)

# Keep required columns and convert to numeric
required = ['Confirmed','Recovered','Deaths','Active']
existing = [c for c in required if c in consolidated.columns]

if 'State' not in consolidated.columns:
    # rename state_col to 'State'
    consolidated = consolidated.rename(columns={state_col: 'State'})

# Convert each existing numeric column safely (remove commas, coerce)
for c in existing:
    consolidated[c] = (consolidated[c]
                       .astype(str)
                       .str.replace(',', '', regex=False)
                       .replace({'nan': None, 'None': None}))
    consolidated[c] = pd.to_numeric(consolidated[c], errors='coerce').fillna(0).astype(int)

# Recompute derived columns
if set(['Confirmed','Recovered','Deaths']).issubset(consolidated.columns):
    consolidated['Active'] = consolidated['Confirmed'] - consolidated['Recovered'] - consolidated['Deaths']
    consolidated['CFR'] = (consolidated['Deaths'] / consolidated['Confirmed'] * 100).round(2)

# Replace df with cleaned consolidated version
df = consolidated.copy()

print("\nPreview after consolidation and conversion:")
print(df[['State'] + existing + ['CFR']].head(8).to_string())
```

```python
# Visualizations: bar, pie, CFR heatmap, choropleth (robust)
import os
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import json
import pandas as pd

os.makedirs("figures", exist_ok=True)

# assume df already cleaned and available
print("DataFrame shape:", df.shape)
print("States loaded:", len(df['State'].unique()))

# 1) Ensure sorted by Confirmed
df = df.sort_values('Confirmed', ascending=False).reset_index(drop=True)

# 2) Horizontal bar chart: top 10 confirmed
top10 = df.head(10).sort_values('Confirmed', ascending=True)
plt.figure(figsize=(10,6))
sns.set_style("whitegrid")
sns.barplot(x='Confirmed', y='State', data=top10, palette='viridis')
plt.title('Top 10 Indian States by Confirmed COVID-19 Cases (2020-2021)')
plt.xlabel('Confirmed cases')
plt.ylabel('')
plt.tight_layout()
bar_path = 'figures/top10_confirmed_bar.png'
plt.savefig(bar_path, dpi=300)
plt.close()
print("Saved bar chart:", bar_path)

# 3) Pie chart: top 6 share vs others
total_cases = df['Confirmed'].sum()
top6 = df.head(6).copy()
others_sum = total_cases - top6['Confirmed'].sum()
labels = top6['State'].tolist() + ['Others']
sizes = top6['Confirmed'].tolist() + [others_sum]

plt.figure(figsize=(8,8))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title('Share of Confirmed Cases: Top 6 States vs Others (2020-2021)')
pie_path = 'figures/pie_top6_share.png'
plt.savefig(pie_path, dpi=300)
plt.close()
print("Saved pie chart:", pie_path)

# 4) CFR heatmap (vertical list)
plt.figure(figsize=(8, max(6, len(df)/3)))  # height scales with number of states
cfr_df = df[['State','CFR']].set_index('State').sort_values('CFR', ascending=False)
sns.heatmap(cfr_df.T, annot=True, fmt=".2f", cmap='Reds', cbar_kws={'label':'CFR (%)'})
plt.title('Case Fatality Rate (CFR) by State (2020-2021)')
plt.yticks([],[])
heatmap_path = 'figures/cfr_heatmap.png'
plt.tight_layout()
plt.savefig(heatmap_path, dpi=300)
plt.close()
print("Saved CFR heatmap:", heatmap_path)

# 5) Choropleth map (uses data/india_states.geojson)
geojson_path = 'data/india_states.geojson'
choropleth_path = 'figures/choropleth_states_confirmed.png'
choropleth_html = 'figures/choropleth_states_confirmed.html'

if not os.path.exists(geojson_path):
    print("\n⚠️ GeoJSON not found:", geojson_path)
    print("To create the choropleth you must download a valid India states GeoJSON and save it as data/india_states.geojson")
    print("Common sources: 'https://github.com/geohacker/india' or 'https://github.com/vinitshahdeo/Statewise-COVID-Data' (search for states geojson).")
else:
    # load geojson and discover its state name property
    with open(geojson_path, 'r', encoding='utf-8') as f:
        india_geo = json.load(f)

    # Try to identify likely property names containing the state name
    # Inspect the first feature's properties for keys
    props0 = india_geo['features'][0]['properties']
    print("\nGeoJSON properties example keys:", list(props0.keys()))
    # Blindly try common keys in order
    possible_name_keys = ['st_nm', 'name', 'NAME_1', 'STATE_NAME', 'ST_NM', 'state', 'STATE']
    found_key = None
    for k in possible_name_keys:
        if k in props0:
            found_key = k
            break
    if found_key is None:
        # choose first string-like property as fallback
        for k,v in props0.items():
            if isinstance(v, str):
                found_key = k
                break

    print("Using GeoJSON property for state name:", found_key)

    # Show list of geojson state names for mapping convenience (first 20)
    geo_names = [feat['properties'].get(found_key, '').strip() for feat in india_geo['features']]
    print("Sample GeoJSON state names (first 15):", geo_names[:15])

    # Compare df['State'] unique names against geo_names to detect mismatches
    state_names = df['State'].str.strip().unique().tolist()
    mismatches = [s for s in state_names if s not in geo_names]
    if len(mismatches) > 0:
        print("\n⚠️ Found state name mismatches between data and GeoJSON.")
        print("Example mismatches (showing up to 20):", mismatches[:20])
        print("You will need to create a mapping dictionary `name_map = {'data_name':'geojson_name', ...}`")
        print("I can help produce a suggested mapping if you want.")
        # but still attempt to create choropleth by best-effort: use locations argument with names that match
    else:
        print("State names match GeoJSON — proceeding to create choropleth.")

    # Attempt to make choropleth (Plotly)
    try:
        # create a copy where 'State' matches geojson naming; user may have fixed mapping
        fig = px.choropleth(df,
                            geojson=india_geo,
                            featureidkey=f'properties.{found_key}',
                            locations='State',
                            color='Confirmed',
                            color_continuous_scale="OrRd",
                            projection='mercator',
                            title='Statewise Confirmed COVID-19 Cases (2020-2021)')
        fig.update_geos(fitbounds="locations", visible=False)
        # Save static image (requires kaleido); fallback: save html
        try:
            fig.write_image(choropleth_path, scale=2)
            print("Saved choropleth PNG:", choropleth_path)
        except Exception as e_img:
            print("Could not write PNG (kaleido may be missing). Saving interactive HTML instead.")
            fig.write_html(choropleth_html)
            print("Saved choropleth HTML:", choropleth_html)
    except Exception as e:
        print("Error creating choropleth:", e)
        print("You might need to adjust State names to match GeoJSON properties or provide a mapping.")

print("\nDone — all available figures saved to ./figures/")
```
